```python
import numpy as np
from pydtmc import MarkovChain, plot_graph, plot_sequence, plot_redistributions

# Transition matrix (rows sum to 1)
P = np.array([
    [0.70, 0.20, 0.10],  # Sunny -> (Sunny, Cloudy, Rainy)
    [0.30, 0.50, 0.20],  # Cloudy -> ...
    [0.20, 0.40, 0.40],  # Rainy -> ...
])

states = ["Sunny", "Cloudy", "Rainy"]

# 1) Build the Markov chain
mc = MarkovChain(P, states)  # MarkovChain(p, states=...) :contentReference[oaicite:2]{index=2}

# 2) Plot the transition graph (built-in)
plot_graph(mc, edges_label=True, nodes_color=True, nodes_shape=True, dpi=300)

# 3) Simulate a random walk (built-in)
path = mc.simulate(steps=30, initial_state="Sunny", seed=7)  # simulate(...) :contentReference[oaicite:3]{index=3}
print("Simulated path:", path)

# 4) Plot a random walk directly (built-in)
plot_sequence(mc, steps=80, initial_state="Sunny", plot_type="histogram", seed=7)  # plot_sequence(...) :contentReference[oaicite:4]{index=4}

# 5) Plot how the distribution evolves over time (redistribution plot)
plot_redistributions(mc, redistributions=40, initial_status="Sunny", plot_type="projection")  # plot_redistributions(...) :contentReference[oaicite:5]{index=5}

# 6) Stationary distribution(s)
# mc.pi returns a list of stationary distributions (one if ergodic). :contentReference[oaicite:6]{index=6}
print("Stationary distributions:", mc.pi)
```
